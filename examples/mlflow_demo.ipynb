```python
# For the demo we construct tiny lightweight objects rather than rely on untyped legacy namespaces
from pathlib import Path
from biolm_utils.mlflow_integration import start_mlflow_run

# Setup a tiny, explicit object with the minimal attributes the demo needs (structured config is preferred via load_config)
class Settings:
    def __init__(self, mlflow):
        self.mlflow = mlflow

class Args:
    def __init__(self, mode, task, settings):
        self.mode = mode
        self.task = task
        self.settings = settings

class Cfg:
    def __init__(self, learning_rate):
        self.learning_rate = learning_rate

args = Args("fine-tune", "regression", Settings({"enabled": True, "tracking_uri": "file:///tmp/mlruns", "experiment_name": "demo-experiment"}))
cfg = Cfg(learning_rate=1e-3)
out = Path('/tmp/mlruns_demo')
out.mkdir(parents=True, exist_ok=True)

# This will attempt to import mlflow lazily — install mlflow if you want to actually use a tracking store
try:
    with start_mlflow_run(out, args, cfg) as ml: 
        if ml is not None:
            ml.log_metrics({"demo_metric": 0.42})
            print('Logged demo metric to MLflow run')
        else:
            print('MLflow not active (disabled in config)')
except Exception as e:
    print('MLflow not available:', e)
```
